```python
from ultralytics import YOLO
import cv2

# Load model (use your YOLO v26 weights or default)     
model = YOLO("yolo11n.pt")  
# Replace with your model if needed: "yolo26n.pt" or "best.pt"

# Open webcam (0 = default camera)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Cannot access webcam")
    exit()

while True:
    ret, frame = cap.read()
    if not ret:
        break

    # Run YOLO inference
    results = model(frame)

    # Annotated frame
    annotated_frame = results[0].plot()

    # Show output
    cv2.imshow("YOLO Webcam", annotated_frame)

    # Press 'q' to exit
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
